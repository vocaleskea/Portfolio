**We are given:**

\[
x =
\begin{bmatrix}
1 \\
0 \\
1
\end{bmatrix},
\quad
y = [1],
\quad
f = \max(0, Z_n)
\]

**Hidden Unit Weights:**  
\[
\begin{bmatrix}
w_{11}=0.2 & w_{12}=-0.3 \\
w_{13}=0.4 & w_{14}=0.1 \\
w_{15}=-0.5 & w_{16}=0.2
\end{bmatrix}
\]

**Output Unit Weights:**  
\[
w_{21}=-0.3, \quad w_{22}=-0.2
\]

**Bias (θ):**  
\[
\theta_1=-0.4, \quad \theta_2=0.2, \quad \theta_3=0.1
\]


```python
import numpy as np

# Input and target output
x = np.array([[1], [0], [1]])
y = np.array([[1]])

# Hidden layer weights (3x2)
W1 = np.array([[0.2, -0.3],
               [0.4,  0.1],
               [-0.5,  0.2]])

# Output layer weights (1x2)
W2 = np.array([[-0.3, -0.2]])

# Biases
theta = np.array([[-0.4], [0.2], [0.1]])
```

```python
# Compute the hidden layer input Z = W1^T * x + theta
Z = np.dot(W1.T, x) + theta[:2]  # We'll use the first two bias terms for hidden units

# Apply ReLU activation function
f = np.maximum(0, Z)

# Compute the output
y_hat = np.dot(W2, f)

# Display intermediate results
print("Hidden Layer Input (Z):\n", Z)
print("\nActivated Output (f):\n", f)
print("\nPredicted Output (y_hat):\n", y_hat)
```

```python
# Compute error
E = 0.5 * (y - y_hat)**2
print("Error (E):", E[0][0])
```


### **Result Summary**

* **Hidden layer input (Z):** Obtained through matrix multiplication followed by adding the bias term.
* **Activation output (f):** Generated by applying the ReLU activation function to the hidden layer input.
* **Predicted output (ŷ):** Derived using the output layer’s weight calculations.
* **Error (E):** Determined using the formula ( E = \frac{1}{2}(y - \hat{y})^2 ).
